```python
#8. Find the Most Expensive Item
# Collect (name, price) pairs
p = []
for i in data:
    for j in i.get('menuItems') or []:
        for k in j['customConfigs']:
            p.append((j['name'], k['itemPrice']))

# Find the most expensive item
print(max(p,key = lambda x : x[1]))
# most_expensive = max(p, key=lambda x: x[1])
# print(most_expensive)

#print("Most Expensive Item:", most_expensive[0],"," "Price:", most_expensive[1])
```

```python
# catogorizing based on price.
price_ranges = {"Cheap": [], "Moderate": [], "Expensive": []}

for x in data:
    for y in x["menuItems"] or []:
        for c in y["customConfigs"]:
            price = c["itemPrice"] / 100
            name = y["name"]
            if price < 8:
                price_ranges["Cheap"].append(name)
            elif 8 <= price <= 12:
                price_ranges["Moderate"].append(name)
            else:
                price_ranges["Expensive"].append(name)

for k in price_ranges:
    price_ranges[k] = sorted(set(price_ranges[k]))

print("Cheap:", price_ranges["Cheap"])
print("Moderate:", price_ranges["Moderate"])
print("Expensive:", price_ranges["Expensive"])
```

```python
# 10.create simple menu dictionary.
dict_menu = {}
for i in data:
    category = i['name']
    items = []
    for j in i['menuItems'] or []:
        items.append(j['name'])
    dict_menu[category] = items

print(dict_menu)
```
