```python
def add(a,b):
    return a+b

def add(a,b,c):
    return a+b+c

res = add(2,3)
print(res)
```

```python
# to achive the method overloading in python we use *args.
def add(a,b):
    return a+b

def add(*args):
    print(args)

res = add(2,3, "python","ai")
print(res)


# results in tuple (2, 3, 'python', 'ai')
```

```python

def add(a,b):
    return a+b

def add(*args):
    s = 0
    output = ""
    for i in args:
        if type(i) == int:
            s += i
        elif type(i) == str:
            output += i
    return s,output
res = add(2,3, "python","ai")
print(res)
```

```python
def default_args(a =10,b=20):
    return a,b
res = default_args()
# if we pass parameters it will override the default values
res = default_args(5,15)
print(res)
```

```python
def key_word_args(name,age):
    return f'your name is {name} and age is {age}'

#if we dont want to follow the order of parameters we can use key word arguments.


res = key_word_args("kamal",24)
print(res)
```

```python
def key_word_args(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")

#if we dont want to follow the order of parameters we can use key word arguments.


res = key_word_args(name = "kamal", age = 24)
print(res)
```

```python
def demo_args(a,*args,b = 10,**kwargs):
    print(a)
    print(args)
    print(b)
    print(kwargs)
res = demo_args(2,3,4,b = 5, name="kamal", age=24)
print(res)
```

Generator

```python
'''
In Python, a generator is a special type of function that lets you produce a sequence of values one at a time, instead of computing them all at once.

🔑 Key Points about Generators
Definition: A function that uses the yield keyword to return values one by one.

Use: They allow you to iterate through large datasets or infinite sequences without storing everything in memory.

Why we use them:

Memory efficiency → Only one item is kept in memory at a time.

Lazy evaluation → Values are generated only when needed.

Readable code → Easier to write than manually managing iterators.'''
```

```python
def generator(n):
    for i in range(n):
        return i

g = generator(10)
print(g)
# this will print only 0 because return statement will exit the function after first iteration.
```

```python
def generator(n):
    for i in range(n):
        yield i
    
# yield statement will return a value and pause the function, maintaining its state for the next call.
g = generator(10)
for i in g:
    print(i)
    
# this will return values from 0 to 9  one ny one. 
```

scope of functions

```python
'''availability of a variable in a program is called scope. There are mainly 4 types of scopes in python.
1. Local Scope: Variables defined within a function are local to that function and cannot be accessed outside of it.
2. Enclosing Scope: Variables in the local scope of enclosing functions can be accessed by nested functions.
3. Global Scope: Variables defined at the top level of a script or module are global and can be accessed from anywhere within that module.
     they can be accessed using the global keyword inside functions.
4. Built-in Scope: Python has a set of built-in functions and names that are always available.
     These can be accessed from any scope without any special declaration. set of built-in functions and names that are always available.
     These can be accessed from any scope without any special declaration.'''
```

```python
x=10
def display():
    print(x)
display()
print(x)
```

```python
x=10
def display():
    y =20
    print(x)
    print(y)
display()
print(x)
print(y)  # y is in local scope
```

```python
x=10
def display():
    x =20
    y = 30
    print(x)
    print(id(x))
    print(y)
display()
print(x)
print(id(x))
print(y)
```

```python
x=10
def display():
    global x
    x =20
    y = 30
    print(x)
    print(id(x))
    print(y)
display()
print(x)
print(id(x))
print(y)
```

```python
#non-local/enclosed.
#nested function:
def display():
    print("display")
    def innerfunc():
        print("inner function")
        
    innerfunc()
display()
```

```python
def display():
    n =10
    def innerfunc():
        print(n)
    innerfunc()   # x is non-local scope of innerfunc
display()
print(n)
```

```python
def display():
    n =10
    def innerfunc():
        nonlocal n
        n =20
        m =30
        print(n)
        print(m)
        print(id(n))
    innerfunc()
    print(n)
    print(id(n))   # x is non-local scope of innerfunc
display()
```
