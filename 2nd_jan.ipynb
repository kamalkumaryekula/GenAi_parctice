```python
def prime_factors_sum(num):
    s = 0
    i = 2
    while num > 1:
        if num % i == 0:
            s += i
            while num % i == 0:
                num //= i
        i += 1
    return s

n = int(input("Enter a number: "))
if prime_factors_sum(n) == prime_factors_sum(n+1):
    print(n, "is a Ruth-Aaron number")
else:
    print(n, "is NOT a Ruth-Aaron number")
```

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

lower = int(input("Enter lower limit: "))
upper = int(input("Enter upper limit: "))

primes = [p for p in range(lower, upper+1) if is_prime(p)]

row = 1
index = 0
while index < len(primes):
    for j in range(row):
        if index < len(primes):
            print(primes[index], end=" ")
            index += 1
        else:
            print("#", end=" ")
    print()
    row += 1
```
