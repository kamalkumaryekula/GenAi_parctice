```python
i =0
while i <5:              # memory is allocated in each iteration 
    if i == 3:
        break
    else:
        print(i) 
```

```python
for i in range(0,6):        # 1. memory is allocated in the bigging block
    if i == 3:
        break
    else:
        print(i)          # for is more fasterthan while loop beacause of point 1
```
