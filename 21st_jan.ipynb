```python
s = "abcfghxyzabc"
max_str = ""
cur_str = ""
i =0
max_len = 0
max_subs = []

while i < len(s):
  if i > 0 and ord(s[i]) == ord(s[i-1]) + 1:
    cur_str += s[i]
  else:
    cur_str = s[i]
  if len(cur_str) > max_len:
    max_len = len(cur_str)
    max_subs = [cur_str]
  elif len(cur_str) == max_len:
    if cur_str not in max_subs:
      max_subs.append(cur_str)
  i += 1

max_str = max_subs

print(max_str)
```

```python
#s = "abceaxyzbij"
s = input("Enter a string: ")
result = ""
for i in range(len(s)):
    if s[i] in s[i+1:]:
        continue
    result += s[i]

print(result)
```
